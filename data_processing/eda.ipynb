```python
import pandas as pd
import duckdb
```

```python
# input path
mbsf_denom_prefix='../data/symlinks/mbsf_medpar_denom/mbsf_medpar_denom'
```

```python
query = f"""
    SELECT DISTINCT
        bene_id,
        race, 
        sex
    FROM 
        '{mbsf_denom_prefix}_*.parquet'
    WHERE  
        race in ('1', '2') AND
        sex in ('1', '2')
    """
```

```python
## connect to duckdb
conn = duckdb.connect()

print("## obtain unique beneficiaries ----")
df = conn.execute(query).fetchdf()
```

```python
# obtain total number of unique beneficiaries by race
df.groupby('race')['bene_id'].nunique()
```
